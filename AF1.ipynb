### Ejercicio 1: Evaluador de Rendimiento de Modelos

- **Conceptos aplicados:** `def`, `input`, `variables`, operadores aritméticos/lógicos, `if`, `elif`.

**Contexto:** Al evaluar un modelo de clasificación, es común categorizar su exactitud (*Accuracy*) para saber si el modelo es aceptable, requiere ajustes o necesita un rediseño total.

**Instrucción:**
En una celda de código, crea una función llamada `evaluar_modelo()` que solicite al usuario mediante la función `input()` el valor de la exactitud del modelo (un número flotante entre $0.0$ y $1.0$). La función debe utilizar estructuras condicionales (`if`, `elif`, `else`) y operadores lógicos para devolver y mostrar un texto con la categoría correspondiente:

- **Excelente:** Si la exactitud es mayor o igual a $0.90$.
- **Aceptable:** Si está entre $0.75$ (inclusive) y $0.90$.
- **Requiere Ajuste:** Si está entre $0.50$ (inclusive) y $0.75$.
- **Crítico / No apto:** Si es menor a $0.50$.

*(Recuerda invocar la función al final de la celda para probar la interacción).*


```python
def evaluar_modelo(evaluacion):
    if evaluacion >= .90:
        print("Excelente")
    elif evaluacion>= .75 and evaluacion<.90:
        print("Aceptable")
    elif evaluacion>= .5 and evaluacion<.75:
        print("Requiere ajustes")
    elif evaluacion<.5:
        print("Critico no apto")
    else:
        print("Fuera de rango")
evaluar_modelo(.9)
```

### Ejercicio 2: Procesamiento de Hiperparámetros y Lotes

- **Conceptos aplicados:** `def`, `listas`, `tuplas`, bucle `for`, operadores.

**Contexto:** En el entrenamiento de redes neuronales, se evalúan distintas tasas de aprendizaje (*learning rates*) y configuraciones de arquitectura fijas.

**Instrucción:**
En una nueva celda de código:

1. Define una **tupla** global llamada `CONFIGURACION_BASE` que almacene los valores fijos de dimensiones de entrada y salida de una capa (por ejemplo: `(512, 10)`). Esta tupla no debe modificarse.
2. Crea una función llamada `procesar_tasas_aprendizaje()` que reciba una **lista** de tasas de aprendizaje flotantes (ejemplo: `[0.1, 0.01, 0.001, 0.0001]`).
3. Utiliza un bucle `for` para iterar sobre la lista, aplicar un operador aritmético (por ejemplo, multiplicar cada tasa por un factor de escala de $2.0$) y almacenar los nuevos valores en una nueva lista.
4. La función debe retornar e imprimir tanto la nueva lista de tasas escaladas como los valores de la tupla base.

```python
configuracion_base = (512,10)
def procesar_tasas_aprendizaje(l_tasas_aprendizaje):
    operador_factor = 2
    lista_nueva = []
    for i in l_tasas_aprendizaje:
        lista_nueva.append(i * operador_factor)
    print(f'lista inicial {l_tasas_aprendizaje} lista final {lista_nueva}')
    return lista_nueva
procesar_tasas_aprendizaje([0.1, 0.01, 0.001, 0.0001])
#no entiendo muy bien si desea evaluar la tupla o una lista x, entonces ejecuto dos veces para ambos
procesar_tasas_aprendizaje(configuracion_base)
```

### Ejercicio 3: Simulador de Convergencia por Épocas

- **Conceptos aplicados:** `def`, bucle `while`, bucle `for`, operadores, `if`, `input`.

**Contexto:** El entrenamiento de un algoritmo de aprendizaje automático ocurre por épocas (*epochs*), donde el error debe disminuir gradualmente hasta cumplir con un criterio de parada.

**Instrucción:**
En una celda de código adicional, crea una función llamada `simular_entrenamiento()` que cumpla con los siguientes requisitos:

1. Pida al usuario mediante `input()` el número máximo de épocas a simular (un número entero).
2. Utiliza un bucle (`for` o `while`) para simular el avance de las épocas desde la época $1$ hasta el límite indicado.
3. En cada época, simula una pérdida (*loss*) decreciente utilizando operaciones matemáticas con la variable de control del bucle.
4. Utiliza una condición `if` dentro del bucle para verificar si la pérdida simulada cae por debajo de un umbral de tolerancia (por ejemplo, menor a $0.05$). Si esto ocurre, imprime un mensaje indicando que el modelo ha convergido anticipadamente y detén el bucle.

```python
def simular_entrenamiento():
  loss = 1
  learning_rate = 0.01
  epocas = int(input("cuantas epocas se desea simular? "))
  for i in range(epocas):
    if loss > 0.05:
      loss -= learning_rate
    else:
      print("el modelo ha convergido")
      break
    print(loss)
simular_entrenamiento()

```
