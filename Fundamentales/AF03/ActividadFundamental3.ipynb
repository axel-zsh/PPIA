```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score, classification_report

# URL pública del dataset de Diabetes
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv"
columnas = [
    'Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 
    'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age', 'Outcome'
]

df = pd.read_csv(url, header=None, names=columnas)
print("Primeras filas del dataset original:")
print(df.head())
```

```python
print('Resúmen estadístico completo:')
df.describe()
```

```python
print('Tipos de datos:')
print(df.info())
print('\nValores nulos:')
print(df.isnull().sum())
```

```python
Diabeticos = df[df['Outcome'] == 1]
NoDiabeticos = df[df['Outcome'] == 0]
```

```python
plt.scatter(Diabeticos['Glucose'], Diabeticos['BMI'], color='red', alpha=0.5)
plt.scatter(NoDiabeticos['Glucose'], NoDiabeticos['BMI'], color='blue', alpha=0.5)
plt.title("Gráfico de Dispersión")
plt.xlabel("Glucosa")
plt.ylabel("IMC")
plt.show()
```
